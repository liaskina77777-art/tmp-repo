```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

# Для воспроизводимости
np.random.seed(42)
random.seed(42)

# --- ПАРАМЕТРЫ ---
n_rows = 500
start_date = datetime(2023, 1, 1, 0, 0, 0)
end_date = datetime(2023, 12, 31, 23, 0, 0)

# --- БАЗОВЫЕ ДАННЫЕ ---
boroughs = ['Manhattan', 'Brooklyn', 'Queens', 'Bronx', 'Staten Island', 'Airport']
holidays = ['2023-01-01', '2023-01-16', '2023-02-20', '2023-05-29', '2023-06-19',
            '2023-07-04', '2023-09-04', '2023-10-09', '2023-11-23', '2023-12-25']

# --- ГЕНЕРАЦИЯ ДАТ ---
dates = [start_date + timedelta(hours=i) for i in range(n_rows)]
date_strs = [d.strftime('%Y-%m-%d %H') for d in dates]
months = [d.month for d in dates]
hours = [d.hour for d in dates]

# --- ПРИЗНАК ВЫХОДНОГО/ПРАЗДНИКА ---
is_holiday = []
for d in dates:
    date_key = d.strftime('%Y-%m-%d')
    if date_key in holidays or d.weekday() >= 5:  # суббота/воскресенье
        is_holiday.append('Y')
    else:
        is_holiday.append('N')

# --- РАЙОНЫ (с разной интенсивностью) ---
weights = [0.35, 0.25, 0.20, 0.10, 0.05, 0.05]  # Manhattan самый активный
borough_data = np.random.choice(boroughs, size=n_rows, p=weights)

# --- ПОГОДНЫЕ ДАННЫЕ (реалистичные сезонные вариации) ---
# Температура (F): зима ~30-40, лето ~70-85
temp_base = 50 + 25 * np.sin(np.array(months) * 2 * np.pi / 12 - np.pi/2)
temp = temp_base + np.random.normal(0, 10, n_rows)
temp = np.clip(temp, 10, 100).round(1)

# Точка росы
dewp = temp - np.random.uniform(5, 20, n_rows)
dewp = np.clip(dewp, 0, 80).round(1)

# Скорость ветра (mph) - сезонные колебания
wind_base = 10 + 5 * np.sin(np.array(months) * 2 * np.pi / 12)
spd = np.random.gamma(2, 3, n_rows) + wind_base
spd = np.clip(spd, 0, 35).round(1)

# Видимость (мили)
vsb = np.random.gamma(3, 3, n_rows)
vsb = np.clip(vsb, 0.5, 15).round(1)

# Давление (атмосферное)
slp = np.random.normal(29.9, 0.3, n_rows)
slp = np.clip(slp, 29.0, 30.5).round(2)

# Осадки (с вероятностью 30% что есть)
pcp_01 = np.zeros(n_rows)
pcp_06 = np.zeros(n_rows)
pcp_24 = np.zeros(n_rows)

for i in range(n_rows):
    if np.random.random() < 0.30:  # 30% вероятность осадков
        pcp_01[i] = round(np.random.exponential(0.05), 3)  # часовые
        pcp_06[i] = round(pcp_01[i] + np.random.exponential(0.1), 3)
        pcp_24[i] = round(pcp_06[i] + np.random.exponential(0.2), 3)

# --- КОЛИЧЕСТВО ПОЕЗДОК (зависит от погоды, района, времени) ---
pickups = []
for i in range(n_rows):
    # Базовое количество поездок
    base = 50
    
    # Поправка на район
    borough_factor = {
        'Manhattan': 2.0,
        'Brooklyn': 1.5,
        'Queens': 1.2,
        'Bronx': 0.9,
        'Staten Island': 0.5,
        'Airport': 0.7
    }
    base *= borough_factor[borough_data[i]]
    
    # Поправка на час (пик: 7-9, 17-19)
    hour = hours[i]
    if 7 <= hour <= 9:
        base *= 1.5
    elif 17 <= hour <= 19:
        base *= 1.6
    elif 22 <= hour or hour <= 5:
        base *= 0.4
    
    # Поправка на праздник (меньше поездок в праздники)
    if is_holiday[i] == 'Y':
        base *= 0.7
    
    # Поправка на погоду (плохая погода → меньше поездок)
    if temp[i] < 20 or temp[i] > 90:
        base *= 0.7
    if pcp_01[i] > 0.1:
        base *= 0.8
    if spd[i] > 20:
        base *= 0.85
    if vsb[i] < 2:
        base *= 0.7
    
    # Добавляем шум
    pickups.append(int(max(1, base + np.random.normal(0, base * 0.2))))

# --- СОЗДАНИЕ DATAFRAME ---
taxi = pd.DataFrame({
    'pickup_dt': [d.strftime('%Y-%m-%d %H:00:00') for d in dates],
    'pickup_month': months,
    'borough': borough_data,
    'pickups': pickups,
    'hday': is_holiday,
    'spd': spd,
    'vsb': vsb,
    'temp': temp,
    'dewp': dewp,
    'slp': slp,
    'pcp_01': pcp_01,
    'pcp_06': pcp_06,
    'pcp_24': pcp_24
})

# Сохраняем в CSV
taxi.to_csv('taxi_nyc_data.csv', index=False)

print("✅ Датсет создан!")
print(f"📊 Размер: {taxi.shape[0]} строк, {taxi.shape[1]} колонок")
print("\n📋 Первые 5 строк:")
print(taxi.head())

print("\n📊 Статистика:")
print(taxi[['pickups', 'temp', 'spd', 'vsb']].describe())
```
