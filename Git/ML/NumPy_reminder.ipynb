Какие могут быть проблемы при реализации последней функции?

````python
a = np.linspace(-10, 10, 1000)
a
````

````python
import numpy as np

def relu(x):
    """Rectified Linear Unit"""
    return np.maximum(0, x)

def relu6(x):
    """ReLU6: ReLU с ограничением сверху значением 6"""
    return np.minimum(np.maximum(0, x), 6)

def softplus_naive(x, beta=1.0):
    """Наивная реализация SoftPlus (может вызывать проблемы!)"""
    return (1.0 / beta) * np.log(1.0 + np.exp(beta * x))

def softplus_stable(x, beta=1.0):
    """Стабильная реализация SoftPlus"""
    # Используем np.logaddexp для численной стабильности
    return (1.0 / beta) * np.logaddexp(0.0, beta * x)
````

````python
def margginalize_rows(joint_dist):
  return np.sum(joint_dist, axis =1)
````

````python
def cond_prob(joint_dist):
  row_marg= margginalize_rows(joint_dist)
  row_marg_safe=np.where(row_marg == 1, 1e-10, row_marg)
  return joint_dist/row_marg_safe[:, np.newaxis]
````

````python
import numpy as np

def marginalize_rows(joint_dist):
    """
    Маргинализация по строкам (суммирование по столбцам)
    Возвращает маргинальное распределение по строкам
    """
    return np.sum(joint_dist, axis=1)

def conditional_probabilities_rows(joint_dist):
    """
    Условные вероятности по строкам P(столбец|строка)
    """
    # Получаем маргинальные вероятности строк
    row_marginals = marginalize_rows(joint_dist)

    # Для избежания деления на ноль
    # Добавляем маленькое значение к нулевым маргиналам
    row_marginals_safe = np.where(row_marginals == 0, 1e-10, row_marginals)

    # Деление каждого элемента на сумму его строки
    # keepdims=True сохраняет размерность для правильного деления
    return joint_dist / row_marginals_safe[:, np.newaxis]
````

#### Задача 6

Как можно получить квадратную матрицу (произвольного размера) с единицами по обеим диагоналям?

_Указание_: сгенерируйте единичную матрицу функцией `np.eye`. Как можно ее "развернуть"?

````python
def double_eye(n):
    main_diag = np.eye(n, dtype=int)
    anti_diag = np.fliplr(main_diag)  # Отражаем слева направо

    return main_diag | anti_diag
    raise NotImplementedError
````

````python
def double_eye2(n):
  eye = np.zeros((n,n), dtype = np.int64)
  eye[range(n), range(n)] = 1
  eye[range(n), range(n-1,-1,-1)] = 1
  return eye
````
